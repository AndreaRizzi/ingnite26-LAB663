## 0.3 Best result, or fastest result?

Most MAI models come in two versions: a **standard** one and a **Flash** one. Flash does the same job, faster and cheaper.

So the question we ask again and again is the same one the core lab asks about models: *which of our jobs actually need the big one?*

| Pick Flash when… | Pick the standard model when… |
| --- | --- |
| You need lots of results | It's the hero result people will study |
| Someone is waiting, like on a phone line | Quality is the whole point, like a radio ad |
| You're exploring and drafting | It's the final version |

We won't take that table on trust. Every notebook measures both versions on the same job, and you decide.

## 0.6 Is my setup ready?

This cell doesn't call Azure. It only checks which settings are in place, so you know which notebooks you can run live. It prints ✅ or ⚠️, never the values themselves.

> 🔮 **Guess first:** if you haven't run `deploy.sh`, which notebooks can you still read?

```python
import os
import shutil
from pathlib import Path

from dotenv import load_dotenv

HERE = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "labs" / "more").is_dir()) / "labs" / "more"
load_dotenv(HERE.parents[1] / ".env")


def has(name: str) -> bool:
    value = os.environ.get(name, "").strip()
    return bool(value) and "<" not in value


mode = os.environ.get("DEMO_MODE", "replay").strip().lower() or "replay"
print(f"Mode in the repo's .env: {'🔁 replay: reading saved results' if mode == 'replay' else '🎨 live: calling the models'}")
print("Each notebook can override it in its own mode cell.\n")

checks = {
    "01 · Build the launch kit": ["FOUNDRY_PROJECT_ENDPOINT", "MODEL_IMAGE_HERO", "MODEL_IMAGE_BULK"],
    "02 · Give it a voice": ["FOUNDRY_PROJECT_ENDPOINT", "MODEL_VOICE_HERO", "MODEL_VOICE_FAST"],
    "03 · Hear your customers": ["FOUNDRY_PROJECT_ENDPOINT", "MODEL_TRANSCRIBE"],
}
for notebook, names in checks.items():
    missing = [name for name in names if not has(name)]
    saved = len([p for p in (HERE / "examples" / notebook[:2]).glob("*") if p.suffix != ".json"])
    live = "✅ ready to run live" if not missing else f"⚠️ live needs {', '.join(missing)}"
    print(f"{notebook:28} {live:46} · 🔁 {saved} saved results to read")

node = shutil.which("node")
print(f"{'04 · Refresh the storefront':28} {'✅ Node.js found' if node else '⚠️ live needs Node.js':46} · plus GitHub Copilot in VS Code")
```

👀 **What to look for:**

- [ ] Every notebook shows saved results to read, even if nothing is ready to run live.
- [ ] If you ran `deploy.sh`, 01 to 03 show ✅.
- [ ] Nothing secret was printed: no endpoint, no names, no IDs.

⚠️ **The honest limit:** ✅ means the settings exist, not that your sign-in or region works. The first live cell of each notebook is the real test.

## 0.7 Ground rules for every notebook

These keep the labs honest, and they're the same rules the core lab uses.

| Rule | What it means here |
| --- | --- |
| **Show, don't tell** | Every section starts with a customer's question, and answers it by running the model. No feature lists. |
